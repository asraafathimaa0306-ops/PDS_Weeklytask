# Practical 6

```python
# Create a dictionary containing student details
student={"name": "Arrahh","age":18,"course":"Python","mark":85}
print(student)

# Display the original dictionary
print("Original Dictionary:",student)

# Access and display values of specific keys
print("Name:",student["name"])
print("Age:",student["age"])

# Add a new key-value pair
student["college"]="SDNB College"
print("After adding new key:",student)

# Update the value of an existing key
student["mark"]=90
print("After updating mark:",student)

# Remove a key-value pair using pop()
student.pop("age")
print("After removing age:",student)

# Find and display the lemgth 
print("Length of dictionary:",len(student))

# Display all keys
print("Keys:",student.keys())

# Display all keys-value pairs
print("Items:",student.items())

# Traverse the dictionary using for loop
print("Key and Value:")
for key,value in student.items():
    print(key,":",value)

# Check whether a specific key exists
if "name" in student:
    print("Name key exists in the dictionary")
else:
    print("Name key does not exist")

# Create a copy of the dictionary 
student_copy=student.copy()
print("Copied Dictionary:",student_copy)

# Create another dictionary and clear all elements
another_dict={
    "name":"Rehan",
    "age":19,
    "course":"Data Science",
}
print("Another Dictionary:",another_dict)
another_dict.clear()
print("After clear():",another_dict)


# Display updated dictionaries 
print("Updated Student Dictionaries:",student)
print("Copied Dictionary:",student_copy)
```

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print("Array:", a)
print("Sum:", np.sum(a))
print("Mean:", np.mean(a))
print("Median:", np.median(a))
print("Standard Deviation:", np.std(a))
print("Variance:", np.var(a))
print("Minimum:", np.min(a))
print("Maximum:", np.max(a))
print("Range:", np.max(a) - np.min(a))

print("25th Percentile:", np.percentile(a, 25))
print("50th Percentile:", np.percentile(a, 50))
print("75th Percentile:", np.percentile(a, 75))

print("Cumulative Sum:", np.cumsum(a))
print("Sorted Array:", np.sort(a))
```
