# Practical 13

```python
# Grouping And Aggregating Data Using Pandas
import pandas as pd
#creating Dataframe
data = {
    "Department":["HR","IT","HR","IT","SALES","SALES"],
    "Employee":["Amit","Rahul","Priya","Ankit","Sakshi","Riya"],
    "Salary":[30000,50000,35000,60000,45000,40000],
    "Experience":[2,5,3,6,4,2]
}
df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)

# Grouping by Department
grouped=df.groupby("Department")
# Sum Aggregation
print("/nSum of Salary by Department:")
print(grouped["Salary"].sum())
# Mean Aggregation
print("\nAverage salary by department:")

# Maximum Salary
print("\nMaximum salary by department:")
print(grouped["Salary"].max())
# Minimum Salary
print("\nMinimum Salaryby Department:")
print(grouped["Salary"].min())

# Count of employees
print("\nEmployees Count by Department:")
print(grouped["Employee"].count())

# Multiple Aggregations
print("\nMultiple Aggregations:")
print(
    grouped["Salary"].agg(
        ["sum","mean","max","min"]
    )
)

# Grouping Multiple Columns
print("\nGrouping by Department and Experience:")

group_multi=df.groupby(
    ["Department", "Experience"]
)["Salary"].sum()

print(group_multi)

# Iterating  Groups
print("\nDisplay Each Group:")

for name, group in grouped:
    print("\nDepartment:", name)
    print(group)

# Sorting Values
print("\nSorted DataFrame by Salary:")

sorted_df=df.sort_values(by="Salary")

print(sorted_df)
```

```python
# Merge and Join DataFrames using Pandas

import pandas as pd

# -------- CONCATENATE OPERATIONS --------
# (all columns name should be same)

# Creating First DataFrame
df1 = pd.DataFrame({
    "ID": [101, 102, 103, 105],
    "Marks": [95, 88, 90, 85]
}, index=[0, 1, 2, 3])

print("\nDataFrame 1:")
print(df1)


# Creating Second DataFrame
df2 = pd.DataFrame({
    "ID": [201, 202, 203, 205],
    "Marks": [85, 78, 80, 65]
}, index=[4, 5, 6, 7])

print("\nDataFrame 2:")
print(df2)
frames=[df1,df2]
result=pd.concat(frames)
print("concatenate dataframe")
print(result)
# ---------------- MERGE OPERATIONS ----------------
# (at least one column should be same, others can be different)

# Creating First DataFrame
df1 = pd.DataFrame({
    "ID": [101, 102, 103, 104],
    "Name": ["Sakshi", "Rahul", "Ankit", "Priya"]
})

print("DataFrame 1:")
print(df1)


# Creating Second DataFrame
df2 = pd.DataFrame({
    "ID": [101, 102, 103, 105],
    "Marks": [95, 88, 90, 85]
}, index=[0, 1, 2, 3])

print("\nDataFrame 2:")
print(df2)

# ---------------- MERGE OPERATIONS ----------------
# (at least one column should be same, others can be different)

# Creating First DataFrame
df1 = pd.DataFrame({
    "ID": [101, 102, 103, 104],
    "Name": ["Sakshi", "Rahul", "Ankit", "Priya"]
})

print("DataFrame 1:")
print(df1)


# Creating Second DataFrame
df2 = pd.DataFrame({
    "ID": [101, 102, 103, 105],
    "Marks": [95, 88, 90, 85]
}, index=[0, 1, 2, 3])

print("\nDataFrame 2:")
print(df2)
frames = [df1,df2]
result = pd.concat(frames)
print("concatenate dataframes")
print(result)

# ---------------- MERGE OPERATIONS ---------------- ( at least one column should be same, others can be different  )

# Creating First DataFrame
df1 = pd.DataFrame({
    "ID": [101, 102, 103, 104],
    "Name": ["Sakshi", "Rahul", "Ankit", "Priya"]
})

print("DataFrame 1:")
print(df1)

# Creating Second DataFrame
df2 = pd.DataFrame({
    "ID": [101, 102, 103, 105],
    "Marks": [95, 88, 90, 85]
}, index=[0, 1, 2, 3])   

print("\nDataFrame 2:")
print(df2)

# Inner Merge
inner_merge = pd.merge(df1, df2, on="ID", how="inner")

print("\nInner Merge:")
print(inner_merge)

# Left Merge
left_merge = pd.merge(df1, df2, on="ID", how="left")

print("\nLeft Merge:")
print(left_merge)

# Right Merge
right_merge = pd.merge(df1, df2, on="ID", how="right")

print("\nRight Merge:")
print(right_merge)

# Outer Merge
outer_merge = pd.merge(df1, df2, on="ID", how="outer")

print("\nOuter Merge:")
print(outer_merge)

# ---------------- JOIN OPERATIONS ---------------- ( columns names in both datframes is different )

# Creating DataFrames for Join
df3 = pd.DataFrame({
    "Name": ["Amit", "Riya", "Karan"]
}, index=[1, 2, 3])

df4 = pd.DataFrame({
    "Salary": [30000, 40000, 50000]
}, index=[1, 2, 3])

print("\nDataFrame 3:")
print(df3)

print("\nDataFrame 4:")
print(df4)

# Joining DataFrames
joined_df = df3.join(df4)

print("\nJoined DataFrame:")
print(joined_df)

# Joining with Different Join Types
left_join = df3.join(df4, how="left")

print("\nLeft Join:")
print(left_join)

right_join = df3.join(df4, how="right")

print("\nRight Join:")
print(right_join)

outer_join = df3.join(df4, how="outer")

print("\nOuter Join:")
print(outer_join)
```

```python
# Simple plots using Matplotlib

import matplotlib.pyplot as plt

#------------------LINE PLOT--------------------

x=[1,2,3,4,5]
y=[10,20,3,40,50]

plt.figure(figsize=(5,4))

plt.plot(x,y)

plt.title("Line Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.show()

#-------------------BAR GRAPH----------------------
Student=["A","B","C","D"]
marks=[85,90,75,88]

plt.figure(figsize=(5,4))

plt.bar(Student,marks)

plt.title("Bar Graph")
plt.xlabel("Student")
plt.ylabel("Marks")

plt.show()

#---------------------HISTOGRAM-----------------

data=[10,20,20,30,30,30,40,50,50]

plt.figure(figsize=(5,4))
plt.hist(data,bins=5)

plt.title("Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

```python
# Data Visualization using Seaborn
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Creating Sample dataset 
data = {
    "Maths" : [85, 90, 78, 92, 88],
    "Science" : [80, 85, 75, 95, 89],
    "English" : [78, 88, 72, 90, 84]
}

# Creating Dataframe
df= pd.DataFrame(data)
print("Dataset :")
print(df)
#------------------------------HEATMAP-----------------------------
plt.figure(figsize=(5,4))
sns.heatmap(df, annot=True)
plt.title("Heatmap")
plt.show()

#----------------------------BOXPLOT------------------------------

plt.figure(figsize=(5,4))
sns.boxplot(data=df)
plt.title("Boxplot")
plt.show()
#--------------------------PAIR PLOT---------------------------
sns.pairplot(df)
plt.show()
```
